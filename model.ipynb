```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# data import and cleaning

df = pd.read_csv('TrainingData.csv')

final_df = pd.read_csv('ValidationData.csv')
```

```python
df.dropna(inplace=True) # null 1 deger var.
df.shape
df['BUILDINGID'].value_counts()
df['FLOOR'].value_counts()
pd.crosstab(df["BUILDINGID"], df['FLOOR'])

wap_cols = [col for col in df.columns if col.startswith("WAP")]

df_clean = df.copy()

df_clean[wap_cols] = df_clean[wap_cols].replace(100, -110)


X = df_clean[wap_cols]
y = df_clean["FLOOR"]
y_multi = df_clean[["BUILDINGID", "FLOOR"]].astype(int)

```

```python
# model selection KNN first

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier


knn = KNeighborsClassifier(
    n_neighbors=5,
    n_jobs=-1
    )
knn.fit(X, y)

X_validation = final_df[wap_cols].copy()

X_validation = X_validation.replace(100, -110)

final_df.head()


# validation and test split for knn

from sklearn.model_selection import train_test_split

X_val, X_test, y_val, y_test = train_test_split(
    X_validation,
    final_df["FLOOR"],
    test_size=0.5,
    random_state=42
)

y_pred = knn.predict(X_val)
y_tpred = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
val_accuracy = accuracy_score(y_val, y_pred)

print("KNN VALIDATION Accuracy:", val_accuracy)

from sklearn.metrics import f1_score

macro_f1 = f1_score(
    y_val,
    y_pred,
    average="macro"
)

print("Macro F1:", macro_f1)
```

```python
# random forest model

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
rf.fit(X, y)


rf_pred = rf.predict(X_val)

from sklearn.metrics import accuracy_score, f1_score, classification_report
print("Random Forest Validation Accuracy:", accuracy_score(y_val, rf_pred))

print("Random Forest Macro f1:", f1_score(y_val, rf_pred, average="macro"))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

df_clean.head()

X_building = df_clean[wap_cols].copy()
y_building = df_clean["BUILDINGID"].astype(int)

X_building_validation = final_df[wap_cols].copy()
X_building_validation = X_building_validation.replace(100, -110)
y_building_validation = final_df["BUILDINGID"].astype(int)

x_bval, x_btest, y_bval, y_btest = train_test_split(
    X_building,
    y_building,
    test_size=0.2,
    random_state=42,
    stratify=y_building
)

knn_building = KNeighborsClassifier(
    n_neighbors=5,
    n_jobs=-1
)
knn_building.fit(X_building, y_building)

knn_building_pred = knn_building.predict(x_bval)

print(
    "Building KNN Accuracy:",
    accuracy_score(y_bval, knn_building_pred)
)

print(
    "Building KNN Macro F1:",
    f1_score(y_bval, knn_building_pred, average="macro")
)

print(
    classification_report(y_bval, knn_building_pred)
)
```

```python
# Random Forest for Building ID

rf_building = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)
rf_building.fit(X_building, y_building)

rf_building_pred = rf_building.predict(x_bval)

print("Building RF Accuracy:", accuracy_score(y_bval, rf_building_pred))
print("Building RF Macro F1:", f1_score(y_bval, rf_building_pred, average="macro"))

print(classification_report(y_bval, rf_building_pred))
```

```python
# Comparison of models

results = {
    'Floor KNN': 0.8875,
    'Floor RF': 0.9118,
    'Building KNN': accuracy_score(y_bval, knn_building_pred),
    'Building RF': accuracy_score(y_bval, rf_building_pred)
}

for model, acc in results.items():
    print(f"{model}: {acc:.4f}")
```

```python
X = df_clean[wap_cols].copy()
y_multi = df_clean[["FLOOR", "BUILDINGID"]].astype(int)

y_multi.value_counts()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

rf_multi = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)   

rf_multi.fit(X, y_multi)


X_validation = final_df[wap_cols].copy()
X_validation = X_validation.replace(100, -110)

y_validation_multi = final_df[
    ["FLOOR", "BUILDINGID"]
].astype(int)

x_vl, x_tst, y_vl, y_tst = train_test_split(
    X_validation,
    y_validation_multi,
    test_size=0.2,
    random_state=42,
    stratify=y_validation_multi
)

rf_pred = rf_multi.predict(x_vl)
```

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report
import numpy as np


floor_accuracy = accuracy_score(
    y_vl["FLOOR"],
    rf_pred[:, 0]
)

building_accuracy = accuracy_score(
    y_vl["BUILDINGID"],
    rf_pred[:, 1]
)

floor_macro_f1 = f1_score(
    y_vl["FLOOR"],
    rf_pred[:, 0],
    average="macro"
)

building_macro_f1 = f1_score(
    y_vl["BUILDINGID"],
    rf_pred[:, 1],
    average="macro"
)

both_correct = (
    (y_vl["FLOOR"].to_numpy() == rf_pred[:, 0]) &
    (y_vl["BUILDINGID"].to_numpy() == rf_pred[:, 1])
)


both_accuracy = both_correct.mean()

print("Floor Macro F1:", floor_macro_f1)
print("Building Macro F1:", building_macro_f1)

print("Floor Accuracy:", floor_accuracy)
print("Building Accuracy:", building_accuracy)
print("Both Correct Accuracy:", both_accuracy)

```

```python
rf_pred_test = rf_multi.predict(x_tst)

from sklearn.metrics import accuracy_score, f1_score, classification_report

floor_accuracy_test = accuracy_score(
    y_tst["FLOOR"],
    rf_pred_test[:, 0]
)
building_accuracy_test = accuracy_score(
    y_tst["BUILDINGID"],
    rf_pred_test[:, 1]
)   

print("Floor Accuracy Test:", floor_accuracy_test)
print("Building Accuracy Test:", building_accuracy_test)
print("Floor Macro F1 Test:", f1_score(y_tst["FLOOR"], rf_pred_test[:, 0], average="macro"))
print("Building Macro F1 Test:", f1_score(y_tst["BUILDINGID"], rf_pred_test[:, 1], average="macro"))
print("Both Correct Accuracy:", both_accuracy)
```
